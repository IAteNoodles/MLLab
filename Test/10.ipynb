```python
import seaborn as sns, matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X, y = load_breast_cancer(return_X_y=True)
X = StandardScaler().fit_transform(X)

km = KMeans(2, random_state=42)
pred = km.fit_predict(X)

pca = PCA(2)
X2 = pca.fit_transform(X)
C = pca.transform(km.cluster_centers_)

sns.scatterplot(x=X2[:,0], y=X2[:,1], hue=pred)
plt.scatter(C[:,0], C[:,1], marker='X', s=200, c='red')
plt.show()
```

```python
import seaborn as sns, matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.metrics import *

X, y = load_breast_cancer(return_X_y=True)
X = StandardScaler().fit_transform(X)

km = KMeans(2, random_state=42)
pred = km.fit_predict(X)

pca = PCA(2)
X2 = pca.fit_transform(X)

# Plot 1: Predicted clusters
sns.scatterplot(x=X2[:,0], y=X2[:,1], hue=pred)
plt.show()

# Plot 2: Actual labels
sns.scatterplot(x=X2[:,0], y=X2[:,1], hue=y)
plt.show()

# Plot 3: Clusters + centroids
C = pca.transform(km.cluster_centers_)
sns.scatterplot(x=X2[:,0], y=X2[:,1], hue=pred)
plt.scatter(C[:,0], C[:,1], marker='X', s=200, c='red')
plt.show()

print(confusion_matrix(y, pred))
print(classification_report(y, pred))
```
