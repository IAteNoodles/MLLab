```python
import pandas as pd

def find_s(f):
    df = pd.read_csv(f)
    print("Training Data:\n", df)

    X,y = df.iloc[:,:-1].values, df.iloc[:,-1].values
    h = ['?'] * len(X[0])

    for x,l in zip(X,y):
        if l.lower() == 'yes':
            h = [v if a=='?' or a==v else '?' for a,v in zip(h,x)]
        print("Current Hypothesis:", h)

    return h

print("\nFinal Hypothesis:", find_s(fr"C:\Users\Noodl\Study Materials\MLLab\training_data.csv"))
```
